```python
# Assignment 1: Superhero Class with Inheritance & Encapsulation

class Superhero:
    def __init__(self, name, power, city):
        self.name = name
        self._power = power
        self.city = city

    def display_info(self):
        print(f"🦸 {self.name} protects {self.city} with the power of {self._power}!")

    def use_power(self):
        print(f"{self.name} uses {self._power}!")

    def get_power(self):
        return self._power

    def set_power(self, new_power):
        self._power = new_power


#  Subclasses demonstrating polymorphism
class FlyingHero(Superhero):
    def use_power(self):
        print(f"{self.name} soars through the sky to save {self.city} with {self._power}!")


class StrengthHero(Superhero):
    def use_power(self):
        print(f"{self.name} lifts a car with their {self._power} strength in {self.city}!")


#  Create superhero instances
hero1 = FlyingHero("SkyBlazer", "Flight", "Metropolis")
hero2 = StrengthHero("IronFist", "Super Strength", "Gotham")

#  Test Superhero behavior
hero1.display_info()
hero1.use_power()

hero2.display_info()
hero2.use_power()

#  Encapsulation test
print(f"\nCurrent power: {hero2.get_power()}")
hero2.set_power("Mega Punch")
print(f"Updated power: {hero2.get_power()}")


# ----------------------------------------------
#  Activity 2: Polymorphism Challenge - Animals

class Animal:
    def move(self):
        print("This animal moves in its own way.")


class Dog(Animal):
    def move(self):
        print("🐕 The dog runs on four legs.")


class Bird(Animal):
    def move(self):
        print("🐦 The bird flies in the sky.")


class Fish(Animal):
    def move(self):
        print("🐟 The fish swims in water.")


# 🐾 Create a list of animals
animals = [Dog(), Bird(), Fish()]

print("\n📦 Polymorphism in action (Animal.move()):")
for animal in animals:
    animal.move()
```
